# Data Preprocessing

```python
#check missing values
titanicData.isnull().sum()
```

```python
# manage missing values as you see pertinent

titanicData = titanicData.drop(columns=["deck"])

```

```python
#check missing values
titanicData.isnull().sum()


```

```python
# manage missing values as you see pertinent
titanicData["age"] = titanicData["age"].fillna(titanicData["age"].mean())
```

```python
# manage missing values as you see pertinent
titanicData["embarked"] = titanicData["embarked"].fillna(titanicData["embarked"].mode()[0])
```

```python
# manage missing values as you see pertinent
titanicData["embark_town"] = titanicData["embark_town"].fillna("non renseigné")
```

```python
#check missing values
titanicData.isnull().sum()

```

```python
# Check for duplicate rows
print(titanicData.duplicated().sum())
```

```python
#Drop duplicates keeping the first occurance
titanicData = titanicData.drop_duplicates()
```

```python
# Check for duplicate rows
print(titanicData.duplicated().sum())

```
